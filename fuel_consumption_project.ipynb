## 2. Background and Mathematical Theory

This section introduces the concepts and mathematical relationships needed to study vehicle fuel consumption. It explains how fuel consumption is expressed, how the amount of fuel used can be estimated, and why driving speed is worth investigating.

### 2.1 Fuel consumption

For cars in many European countries, fuel consumption is commonly expressed in litres per 100 kilometres (L/100 km). This unit describes how many litres of fuel a vehicle uses, on average, to travel 100 kilometres.

A lower L/100 km value means that the vehicle uses less fuel to cover the same distance. For example, a vehicle with a consumption of 5 L/100 km uses less fuel over a 100-kilometre journey than a vehicle with a consumption of 8 L/100 km.

Fuel consumption is an average measure. The actual amount used can vary with driving conditions, so a single value should not be treated as a constant that applies to every situation.

### 2.2 Calculating the amount of fuel used

If the average fuel consumption and journey distance are known, the amount of fuel required can be estimated using:

$$F = \frac{D \times C}{100}$$

Where:

- $F$ is the estimated fuel used, in litres.
- $D$ is the distance travelled, in kilometres.
- $C$ is the average fuel consumption, in litres per 100 kilometres.

The division by 100 is necessary because the consumption value is given for every 100 kilometres.

For example, if a vehicle averages 6 L/100 km over a journey of 250 km, the estimated fuel used is:

$$F = \frac{250 \times 6}{100} = 15\text{ litres}$$

This calculation assumes that the average consumption value is representative of the whole journey.

### 2.3 Calculating fuel cost

Once the amount of fuel has been estimated, the fuel cost can be calculated by multiplying the fuel used by the price per litre:

$$K = F \times P$$

Where:

- $K$ is the estimated fuel cost, in euros.
- $F$ is the amount of fuel used, in litres.
- $P$ is the fuel price, in euros per litre.

For example, if a journey requires 15 litres of fuel and the assumed fuel price is €1.80 per litre, the estimated cost is €27.00.

The fuel price in this example is illustrative. Actual prices vary by location and over time.

### 2.4 Driving speed and fuel consumption

Driving speed is one factor that may influence a vehicle's fuel consumption. The relationship is not necessarily a simple straight line: fuel consumption can differ at low, moderate and high speeds, and the pattern depends on the vehicle and driving conditions.

Other factors can also affect fuel use, including acceleration, braking, road gradient, traffic, weather, vehicle load and engine condition.

For this project, we will focus on driving speed while keeping other conditions as consistent as possible in our data or assumptions. This will help us investigate the effect of speed without claiming that speed is the only factor that affects fuel consumption.

Research supports investigating this relationship. The U.S. Department of Energy summarises an Oak Ridge National Laboratory study of 74 vehicles, reporting that the best fuel economy was generally found around 40–50 mph and that fuel economy was about 27% lower at 80 mph than at 60 mph. These findings describe a broad trend across the vehicles studied; they do not provide measured consumption values for the hypothetical vehicle used in this project ([U.S. Department of Energy, 2022](https://www.energy.gov/cmei/vehicles/articles/fotw-1240-may-30-2022-slow-down-save-fuel-fuel-economy-decreases-about-27)).

The U.S. Environmental Protection Agency also explains that fuel economy varies with factors such as driving behaviour, vehicle condition, weather, road conditions and load ([U.S. EPA, *Your Mileage May Vary*](https://www.epa.gov/greenvehicles/your-mileage-may-vary)). This supports the decision to state simplifying assumptions and limitations clearly.

## 6. Mathematical Modelling

This chapter uses regression analysis to describe the relationship between driving speed and the illustrative fuel-consumption values. Two mathematical models will be fitted and compared: a linear model and a quadratic model.

The linear model represents the relationship with a straight line. The quadratic model includes a squared speed term, allowing the curve to decrease and then increase. This makes it a possible candidate for the U-shaped pattern observed in Chapter 5.

The models are fitted using the least-squares method. This method chooses the coefficients that minimise the sum of the squared differences between the observed values and the model's predictions.

### 6.1 Choosing the Model Types

A linear model has the form:

\[
C(v) = av + b
\]

A quadratic model has the form:

\[
C(v) = av^2 + bv + c
\]

Here, \(v\) represents driving speed in km/h, and \(C(v)\) represents the modelled fuel consumption in L/100 km. The letters \(a\), \(b\) and \(c\) are coefficients calculated from the data.

The linear model is included as a simple baseline. However, a straight line cannot represent a decrease followed by an increase. A quadratic model can represent this shape because its graph is a parabola.

The choice of a quadratic model is therefore motivated by the pattern in the exploratory analysis. It is not assumed to be a universal physical law for vehicle fuel consumption.

### 6.2 Fitting the Linear and Quadratic Models

The `numpy.polyfit()` function calculates polynomial coefficients using least-squares regression. A degree of 1 produces a linear model, while a degree of 2 produces a quadratic model.

The code below fits both models to the same five observations and calculates their predicted consumption values at the recorded speeds.

```python
# Fit a linear model (polynomial degree 1)
linear_coefficients = np.polyfit(speed_kmh, consumption_l_per_100km, 1)
linear_model = np.poly1d(linear_coefficients)

# Fit a quadratic model (polynomial degree 2)
quadratic_coefficients = np.polyfit(speed_kmh, consumption_l_per_100km, 2)
quadratic_model = np.poly1d(quadratic_coefficients)

# Calculate predictions at the observed speeds
linear_predictions = linear_model(speed_kmh)
quadratic_predictions = quadratic_model(speed_kmh)

print("Linear model coefficients:")
print(f"a = {linear_coefficients[0]:.6f}")
print(f"b = {linear_coefficients[1]:.6f}")
print()
print("Quadratic model coefficients:")
print(f"a = {quadratic_coefficients[0]:.8f}")
print(f"b = {quadratic_coefficients[1]:.8f}")
print(f"c = {quadratic_coefficients[2]:.8f}")
```

The linear model has two coefficients, while the quadratic model has three. The quadratic model is more flexible because it can represent curvature.

The coefficients are numerical results calculated from the illustrative dataset. Their values are specific to the five observations used in this project.

### 6.3 Comparing Model Fit

To compare the models, two measures will be calculated:

- **Coefficient of determination (\(R^2\))**: describes the proportion of variation in the observed values accounted for by the model. A value closer to 1 indicates a closer fit to these data.
- **Root mean squared error (RMSE)**: measures the typical size of the prediction errors, in L/100 km. A smaller value indicates that predictions are closer to the observed values.

These measures describe how well each model fits the dataset. They do not prove that a model will accurately predict the fuel consumption of a real vehicle.

```python
def calculate_r_squared(actual, predicted):
    residual_sum_of_squares = np.sum((actual - predicted) ** 2)
    total_sum_of_squares = np.sum((actual - np.mean(actual)) ** 2)
    return 1 - residual_sum_of_squares / total_sum_of_squares

def calculate_rmse(actual, predicted):
    return np.sqrt(np.mean((actual - predicted) ** 2))

model_comparison = pd.DataFrame({
    "Model": ["Linear", "Quadratic"],
    "R-squared": [
        calculate_r_squared(consumption_l_per_100km, linear_predictions),
        calculate_r_squared(consumption_l_per_100km, quadratic_predictions)
    ],
    "RMSE (L/100 km)": [
        calculate_rmse(consumption_l_per_100km, linear_predictions),
        calculate_rmse(consumption_l_per_100km, quadratic_predictions)
    ]
})

model_comparison.round(4)
```

The quadratic model is expected to fit this particular dataset more closely because it can represent the decrease and subsequent increase in consumption. The comparison table provides the numerical evidence for that assessment.

A higher \(R^2\) and lower RMSE indicate a closer fit to the five selected values. However, the quadratic model has an additional coefficient and greater flexibility. With only five illustrative observations, a close fit should not be confused with independent validation or proof that the model describes a real vehicle.

### 6.4 The Quadratic Model Equation

The fitted quadratic model can be written as:

\[
C(v) = av^2 + bv + c
\]

The coefficients are inserted into the equation using the values calculated by Python. The vertex of a quadratic function occurs at:

\[
v_{\text{min}} = -\frac{b}{2a}
\]

For this dataset, the coefficient of \(v^2\) is positive, so the parabola opens upwards and its vertex represents a minimum. The vertex gives the minimum predicted consumption according to the fitted quadratic model.

```python
a, b, c = quadratic_coefficients

vertex_speed = -b / (2 * a)
minimum_predicted_consumption = quadratic_model(vertex_speed)

print(
    f"Quadratic model: C(v) = {a:.8f}v² "
    f"+ ({b:.8f})v + {c:.8f}"
)
print(f"Predicted minimum occurs at approximately {vertex_speed:.2f} km/h.")
print(
    "Predicted consumption at this speed: "
    f"{minimum_predicted_consumption:.2f} L/100 km"
)
```

The fitted quadratic equation estimates its minimum at approximately **73 km/h**, with a predicted consumption of approximately **5.10 L/100 km**.

This is close to the lowest observed data point, which is 5.0 L/100 km at 70 km/h. The small difference occurs because regression finds the curve that minimises the total squared error across all five observations; it does not have to pass exactly through every point.

The estimated minimum is a feature of this fitted equation and this illustrative dataset. It should not be interpreted as a recommended or universally optimal driving speed.

### 6.6 Examining the Residuals

A residual is the difference between an observed value and the value predicted by a model:

\[
e_i = C_i - \hat{C}_i
\]

Here, \(C_i\) is an observed fuel-consumption value and \(\hat{C}_i\) is the corresponding model prediction. A positive residual means the model predicted too little consumption; a negative residual means it predicted too much.

The residual plot helps show where the quadratic model differs from the observations.

```python
quadratic_residuals = consumption_l_per_100km - quadratic_predictions

residual_data = pd.DataFrame({
    "Speed (km/h)": speed_kmh,
    "Observed consumption (L/100 km)": consumption_l_per_100km,
    "Predicted consumption (L/100 km)": quadratic_predictions,
    "Residual (L/100 km)": quadratic_residuals
})

residual_data.round(3)
```

```python
plt.figure(figsize=(8, 5))
plt.axhline(0, color="black", linewidth=0.9)
plt.scatter(speed_kmh, quadratic_residuals, color="tab:purple")

plt.title("Residuals of the Quadratic Fuel-Consumption Model")
plt.xlabel("Driving speed (km/h)")
plt.ylabel("Residual (L/100 km)")
plt.xticks(speed_kmh)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

The residuals show the differences between the five illustrative values and the quadratic model's predictions. Values close to zero indicate that the prediction is close to the corresponding observation.

Because the quadratic model is fitted using least squares, it minimises the sum of squared residuals for this dataset. The residual plot is a useful diagnostic, but five points are too few to establish that the model's errors are random or that the model will perform well on new data.

### 7.1 Mathematical Formulas

Fuel consumption is expressed in litres per 100 kilometres. To estimate the fuel required for a journey, the distance is multiplied by the consumption rate and divided by 100:

\[
F = rac{D 	imes C}{100}
\]

Where:
- \(F\) = fuel required (litres)
- \(D\) = journey distance (km)
- \(C\) = average fuel consumption (L/100 km)

The estimated fuel cost is then calculated by multiplying the required fuel by the price per litre:

\[
K = F 	imes P
\]

Where:
- \(K\) = fuel cost (euros)
- \(P\) = petrol price (euros per litre)

Combining the two equations gives:

\[
oxed{K = rac{D 	imes C 	imes P}{100}}
\]

The fuel-consumption value \(C\) can be estimated using the quadratic model from Chapter 6:

\[
C(v) = av^2 + bv + c
\]

Therefore, the estimated fuel cost for a journey driven at a constant speed can be expressed as:

\[
K(D,v,P) = rac{D 	imes C(v) 	imes P}{100}
\]

This calculation assumes that the selected average speed and fuel-consumption rate remain representative throughout the journey.

## 8. Model Validation and Reliability Assessment

A mathematical model should not only fit the data used to create it; it should also be examined to see how well it predicts values that were not included in the fitting process.

Chapter 6 compared the linear and quadratic models using the original five observations. This chapter takes the evaluation one step further by using **leave-one-out cross-validation**. This method repeatedly fits a model using four observations and tests its prediction against the one observation left out.

The dataset is very small and illustrative, so this process cannot establish how accurately either model predicts the fuel consumption of a real vehicle. It does, however, provide a more informative comparison of how the two equations behave when one of the supplied observations is temporarily withheld.

### 8.1 In-Sample Fit and Predictive Performance

The \(R^2\) and RMSE values calculated in Chapter 6 describe how closely each model fits the same five observations used to calculate its coefficients. This is known as **in-sample performance**.

A flexible model can fit a small dataset very closely, but that does not automatically mean it will make accurate predictions for new observations. To investigate this distinction, the following sections use cross-validation.

In this project, the two models being compared are:

- **Linear model:** a polynomial of degree 1.
- **Quadratic model:** a polynomial of degree 2.

Both models are evaluated using the same observations and the same validation procedure.

### 8.2 Leave-One-Out Cross-Validation

Leave-one-out cross-validation (LOOCV) is a procedure in which each observation is left out once.

With five observations, the procedure has five rounds:

1. Four observations are used to fit the model.
2. The fitted model predicts the observation that was left out.
3. The prediction error is calculated.
4. The process is repeated until every observation has been used as the test observation.

This means that each prediction is made by a model that was not fitted using the value it is trying to predict. The following code performs this procedure for both the linear and quadratic models.

```python
# Store cross-validation predictions for each model.
loocv_predictions = {
    "Linear": np.zeros(len(speed_kmh)),
    "Quadratic": np.zeros(len(speed_kmh))
}

# Leave out one observation at a time.
for i in range(len(speed_kmh)):
    training_mask = np.arange(len(speed_kmh)) != i

    # Training data: all observations except the one being tested.
    speed_train = speed_kmh[training_mask]
    consumption_train = consumption_l_per_100km[training_mask]

    # Fit each model using only the training observations.
    linear_cv_coefficients = np.polyfit(speed_train, consumption_train, 1)
    quadratic_cv_coefficients = np.polyfit(speed_train, consumption_train, 2)

    # Predict the observation that was left out.
    loocv_predictions["Linear"][i] = np.polyval(
        linear_cv_coefficients, speed_kmh[i]
    )
    loocv_predictions["Quadratic"][i] = np.polyval(
        quadratic_cv_coefficients, speed_kmh[i]
    )

print("Leave-one-out cross-validation predictions calculated.")
```

### 8.3 Examining the Cross-Validation Predictions

The table below places each original illustrative value beside the prediction made when that observation was left out of the fitting process.

The prediction error is calculated as:

\[
e_i = C_i - \hat{C}_i
\]

A positive error means that the model predicted a lower consumption value than the illustrative observation. A negative error means that it predicted a higher value.

```python
loocv_data = pd.DataFrame({
    "Speed (km/h)": speed_kmh,
    "Observed consumption (L/100 km)": consumption_l_per_100km,
    "Linear LOOCV prediction (L/100 km)": loocv_predictions["Linear"],
    "Quadratic LOOCV prediction (L/100 km)": loocv_predictions["Quadratic"]
})

loocv_data["Linear error (L/100 km)"] = (
    loocv_data["Observed consumption (L/100 km)"]
    - loocv_data["Linear LOOCV prediction (L/100 km)"]
)

loocv_data["Quadratic error (L/100 km)"] = (
    loocv_data["Observed consumption (L/100 km)"]
    - loocv_data["Quadratic LOOCV prediction (L/100 km)"]
)

loocv_data.round(3)
```

### 8.4 Comparing Predictive Errors

Two error measures are used to compare the cross-validation predictions:

- **Mean Absolute Error (MAE):** the average absolute difference between the observed values and predictions. It is expressed in L/100 km.
- **Root Mean Squared Error (RMSE):** the square root of the average squared prediction errors. It is also expressed in L/100 km and gives greater weight to larger errors.

Lower MAE and RMSE values indicate smaller prediction errors for this cross-validation exercise. These measures are calculated from the predictions made while each observation was left out.

```python
def calculate_mae(actual, predicted):
    return np.mean(np.abs(actual - predicted))

loocv_comparison = pd.DataFrame({
    "Model": ["Linear", "Quadratic"],
    "LOOCV MAE (L/100 km)": [
        calculate_mae(
            consumption_l_per_100km, loocv_predictions["Linear"]
        ),
        calculate_mae(
            consumption_l_per_100km, loocv_predictions["Quadratic"]
        )
    ],
    "LOOCV RMSE (L/100 km)": [
        calculate_rmse(
            consumption_l_per_100km, loocv_predictions["Linear"]
        ),
        calculate_rmse(
            consumption_l_per_100km, loocv_predictions["Quadratic"]
        )
    ]
})

loocv_comparison.round(4)
```

For the five illustrative observations, the quadratic model produces lower leave-one-out MAE and RMSE values than the linear model. This suggests that, within this small dataset, the quadratic model predicts the withheld observations more closely.

The result supports using the quadratic equation for the project's later calculations, but it is not proof that the quadratic model is universally more accurate. The comparison is based on only five assumed values, and a different dataset could produce a different result.

### 8.5 Visualising the Cross-Validation Results

A useful way to compare predictions is to plot each model's cross-validation predictions against the original observations. If a prediction is close to the observed value, its point will be close to the diagonal reference line.

The chart uses the same axes for both models, making it easier to see which set of predictions is closer to the reference line.

```python
minimum_value = min(
    consumption_l_per_100km.min(),
    loocv_predictions["Linear"].min(),
    loocv_predictions["Quadratic"].min()
)
maximum_value = max(
    consumption_l_per_100km.max(),
    loocv_predictions["Linear"].max(),
    loocv_predictions["Quadratic"].max()
)

plt.figure(figsize=(8, 6))

plt.scatter(
    consumption_l_per_100km,
    loocv_predictions["Linear"],
    label="Linear model",
    marker="o"
)
plt.scatter(
    consumption_l_per_100km,
    loocv_predictions["Quadratic"],
    label="Quadratic model",
    marker="s"
)

plt.plot(
    [minimum_value, maximum_value],
    [minimum_value, maximum_value],
    linestyle="--",
    color="black",
    label="Perfect prediction"
)

plt.title("Observed Values vs Leave-One-Out Predictions")
plt.xlabel("Observed fuel consumption (L/100 km)")
plt.ylabel("Cross-validation prediction (L/100 km)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

### 8.6 Limitations of the Validation

Cross-validation provides an additional check, but several limitations remain:

- **Very small dataset:** each training round uses only four observations. A small change in the data can have a noticeable effect on the fitted coefficients.
- **Illustrative values:** the five consumption values are assumptions, not measurements collected from a real vehicle.
- **No independent real-world test:** the model has not been tested against a separate dataset or controlled vehicle measurements.
- **Limited conditions:** the model does not account for traffic, gradients, weather, acceleration, vehicle load or differences between individual vehicles.
- **Limited speed range:** the results only relate to the selected speeds between 50 and 130 km/h. The equations should not be assumed to behave realistically outside that interval.

The cross-validation results therefore compare the two models' behaviour on the project's illustrative dataset. They do not establish real-world predictive accuracy. A stronger validation would require a larger set of measured fuel-consumption observations, ideally collected under controlled and documented driving conditions.

## 9. Scenario Analysis and Practical Applications

The previous chapters used the mathematical model to estimate fuel consumption and road-trip costs, and then evaluated how well the model represents the illustrative dataset.

This chapter investigates how the estimated cost changes when some of the inputs are different. This is called **scenario analysis**. It is useful because the fuel price and the actual fuel consumption of a vehicle may differ from the values used in a calculation.

The analysis uses the quadratic model from Chapter 6 and the same example journey introduced in Chapter 7. The scenarios are hypothetical and are intended to demonstrate how the calculations respond to changing assumptions. They are not forecasts of fuel prices or guarantees of real-world vehicle performance.

### 9.1 Purpose of Scenario Analysis

A road-trip cost estimate depends on several inputs. In this project, the main inputs are:

- Journey distance, measured in kilometres.
- Estimated fuel consumption, measured in litres per 100 kilometres.
- Fuel price, measured in euros per litre.

The fuel cost is calculated using:

\[
K = \frac{D \times C}{100} \times P
\]

where \(D\) is the journey distance, \(C\) is fuel consumption and \(P\) is the fuel price.

Scenario analysis changes one or more of these inputs while keeping the calculation method the same. This makes it possible to see which assumptions have the greatest effect on the estimated cost.

### 9.2 Comparing Different Fuel Prices

Fuel prices can change over time and may differ between petrol stations. To investigate the effect of price, the journey distance and estimated fuel consumption are kept constant while the price per litre is changed.

The example uses:

- Journey distance: **250 km**.
- Constant driving speed for the model estimate: **90 km/h**.
- Assumed fuel prices: **€1.50, €1.80 and €2.10 per litre**.

These prices are selected for comparison only. They are not live market prices.

```python
# Inputs for the fuel-price scenario analysis.
scenario_distance_km = 250
scenario_speed_kmh = 90
scenario_fuel_prices = np.array([1.50, 1.80, 2.10])

# Estimate consumption using the quadratic model.
scenario_consumption = float(quadratic_model(scenario_speed_kmh))

# Calculate the fuel required for the journey.
scenario_fuel_required = (
    scenario_distance_km * scenario_consumption / 100
)

fuel_price_results = pd.DataFrame({
    "Fuel price (€/L)": scenario_fuel_prices,
    "Fuel required (L)": scenario_fuel_required
})

fuel_price_results["Estimated cost (€)"] = (
    fuel_price_results["Fuel required (L)"]
    * fuel_price_results["Fuel price (€/L)"]
)

print(f"Estimated consumption at {scenario_speed_kmh} km/h: "
      f"{scenario_consumption:.2f} L/100 km")
print(f"Fuel required for {scenario_distance_km} km: "
      f"{scenario_fuel_required:.2f} L")

fuel_price_results.round(2)
```

The table shows that the estimated fuel required stays the same because the journey distance and modelled consumption have not changed. Only the cost changes.

The relationship between fuel price and fuel cost is linear in this calculation: if the price per litre increases by a certain percentage, the estimated fuel cost increases by the same percentage, provided that fuel consumption and distance remain constant.

```python
plt.figure(figsize=(8, 5))
plt.plot(
    fuel_price_results["Fuel price (€/L)"],
    fuel_price_results["Estimated cost (€)"],
    marker="o"
)

plt.title("Estimated Journey Cost at Different Fuel Prices")
plt.xlabel("Assumed fuel price (€/L)")
plt.ylabel("Estimated fuel cost (€)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### 9.3 Investigating Uncertainty in Fuel Consumption

The quadratic model provides an estimate, but actual fuel consumption may be higher or lower because of factors that are not represented in the model. Examples include traffic, weather, road gradients, vehicle load and driving behaviour.

To illustrate the effect of this uncertainty, the model's estimated consumption is adjusted by -10%, -5%, 0%, +5% and +10%. The journey distance and fuel price are kept constant.

The percentage adjustments are hypothetical test values. They do not represent a measured error range for the vehicle or a statistical confidence interval.

```python
# Test different percentage adjustments to the modelled consumption.
consumption_adjustments_percent = np.array([-10, -5, 0, 5, 10])

adjusted_consumption = scenario_consumption * (
    1 + consumption_adjustments_percent / 100
)

consumption_sensitivity = pd.DataFrame({
    "Consumption adjustment (%)": consumption_adjustments_percent,
    "Adjusted consumption (L/100 km)": adjusted_consumption
})

consumption_sensitivity["Fuel required (L)"] = (
    scenario_distance_km
    * consumption_sensitivity["Adjusted consumption (L/100 km)"]
    / 100
)

consumption_sensitivity["Estimated cost (€)"] = (
    consumption_sensitivity["Fuel required (L)"]
    * 1.80
)

consumption_sensitivity.round(2)
```

The table demonstrates that fuel required and fuel cost change in direct proportion to the assumed consumption. For example, a 10% increase in consumption produces a 10% increase in fuel required and fuel cost when distance and fuel price remain unchanged.

This is important when interpreting the results: even if the mathematical calculation is correct, an inaccurate consumption estimate will lead to an inaccurate cost estimate.

```python
plt.figure(figsize=(8, 5))
plt.plot(
    consumption_sensitivity["Consumption adjustment (%)"],
    consumption_sensitivity["Estimated cost (€)"],
    marker="o"
)

plt.title("Effect of Consumption Changes on Estimated Journey Cost")
plt.xlabel("Adjustment to modelled consumption (%)")
plt.ylabel("Estimated fuel cost (€)")
plt.xticks(consumption_adjustments_percent)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### 9.4 Combining Different Assumptions

In a real journey, fuel price and fuel consumption may both differ from the baseline assumptions. This section combines the two factors to create three example situations:

- **Optimistic scenario:** fuel is assumed to cost €1.60 per litre and consumption is 5% lower than the model estimate.
- **Baseline scenario:** fuel costs €1.80 per litre and consumption matches the model estimate.
- **Pessimistic scenario:** fuel is assumed to cost €2.00 per litre and consumption is 10% higher than the model estimate.

These labels describe the assumptions used in the calculation. They do not imply that any scenario is certain to occur.

```python
combined_scenarios = pd.DataFrame({
    "Scenario": ["Optimistic", "Baseline", "Pessimistic"],
    "Fuel price (€/L)": [1.60, 1.80, 2.00],
    "Consumption adjustment (%)": [-5, 0, 10]
})

combined_scenarios["Consumption (L/100 km)"] = (
    scenario_consumption
    * (1 + combined_scenarios["Consumption adjustment (%)"] / 100)
)

combined_scenarios["Fuel required (L)"] = (
    scenario_distance_km
    * combined_scenarios["Consumption (L/100 km)"]
    / 100
)

combined_scenarios["Estimated cost (€)"] = (
    combined_scenarios["Fuel required (L)"]
    * combined_scenarios["Fuel price (€/L)"]
)

combined_scenarios.round(2)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(
    combined_scenarios["Scenario"],
    combined_scenarios["Estimated cost (€)"]
)

plt.title("Estimated Cost Under Combined Scenarios")
plt.xlabel("Scenario")
plt.ylabel("Estimated fuel cost (€)")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

The combined scenarios show how changes in more than one input can affect the final estimate. The difference between the optimistic and pessimistic results is not caused by a change in journey distance; it comes from the assumed fuel price and fuel consumption.

This type of comparison can be useful when planning a budget. Instead of relying on one exact figure, a driver can consider a range of possible fuel costs.

- Changes in fuel price and consumption can combine, producing a larger difference in the final estimate.
- Comparing several scenarios can provide a more useful budgeting range than relying on a single estimate.

These findings show that the model can be used as a basic planning tool. However, the estimated costs cover fuel only and do not represent the total cost of a road trip.

### 10.4 Overall Evaluation of the Project

The project successfully connected mathematical theory with a practical transport-related problem. It used Python to organise data, perform calculations, fit regression models, evaluate prediction errors and create visualisations.

Several aspects of the approach were useful:

- The variables and units were defined before the calculations were performed.
- The dataset was inspected and checked for basic data-quality problems.
- More than one mathematical model was considered rather than assuming a particular relationship in advance.
- Model performance was examined using numerical error measures and cross-validation.
- The model was applied to journey-cost calculations and alternative scenarios.
- Assumptions and limitations were discussed so that the results were not presented as more certain than the data allows.

The project also demonstrates the importance of distinguishing between a calculation being mathematically correct and its inputs accurately representing the real world. Python can perform the calculations consistently, but it cannot compensate for unrealistic or insufficient input data.

### 10.5 Limitations of the Research

The most important limitation is that the five fuel-consumption observations were created as illustrative values. They were not collected through controlled vehicle tests or taken from a measured dataset for the hypothetical vehicle.

This affects the strength of the conclusions. The fitted equations describe the selected values, but they cannot establish the actual fuel-consumption behaviour of a particular car.

Other limitations include:

- **Small sample size:** five observations provide limited information about the relationship between speed and fuel consumption.
- **Simplified driving conditions:** the model does not directly account for traffic, acceleration, braking, road gradients, weather, tyre pressure, vehicle load or engine temperature.
- **Constant-speed assumption:** real journeys involve changes in speed and may include stops, congestion and different road types.
- **Vehicle-specific behaviour:** different engines, transmissions, vehicle shapes, weights and maintenance conditions can produce different consumption patterns.
- **Restricted speed range:** the model is based on observations from 50 to 130 km/h and should not be assumed to give realistic results outside this interval.
- **Fuel-only costs:** tolls, parking, maintenance, depreciation and other travel expenses were not included.

Consequently, the project demonstrates a modelling method rather than providing a verified fuel-consumption calculator for a real vehicle.

### 10.6 Recommendations for Further Research

The project could be developed further by replacing the illustrative dataset with measured fuel-consumption data. A future investigation could record fuel use at different steady speeds while documenting the vehicle, route, weather, load and test conditions.

A larger dataset would make it possible to investigate whether the quadratic model remains appropriate or whether another model provides a better representation. Measurements could also be repeated at each speed to examine variation and uncertainty.

Further improvements could include:

- Comparing multiple vehicle types, engine sizes or transmission types.
- Separating motorway, rural-road and urban driving conditions.
- Investigating the effects of vehicle load, temperature and road gradient.
- Using a more detailed journey model that accounts for changes in speed rather than assuming one constant speed.
- Comparing estimated fuel costs with actual fuel purchases from completed journeys.
- Extending the cost calculation to include tolls, parking and other journey expenses.

These improvements would make the model more representative of real transport conditions and could support a more reliable practical application.

### 10.7 Final Conclusion

This project demonstrated how mathematical modelling and Python can be used to investigate the relationship between driving speed and fuel consumption and to estimate the fuel cost of a road trip.

By preparing and exploring an illustrative dataset, fitting linear and quadratic regression models, evaluating prediction errors and applying the selected model to journey calculations, the project established a complete example of a mathematical modelling process.

The results showed how a non-linear model can represent a curved relationship in the selected data and how estimated fuel consumption can be used to calculate fuel requirements and costs. The scenario analysis further demonstrated how changes in fuel price and consumption assumptions affect the final estimate.

However, because the dataset was illustrative and limited to five observations, the model's real-world accuracy cannot be established from this investigation. The numerical results should therefore be understood as demonstrations of the method, not as definitive predictions for an actual vehicle.

**In conclusion, mathematical modelling provides a useful framework for analysing fuel consumption and estimating road-trip fuel costs, but the reliability of its predictions depends on the quality of the data, the assumptions made and how closely the model represents real driving conditions.**
