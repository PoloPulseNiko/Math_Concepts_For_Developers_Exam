## 2. Background and Mathematical Theory

This section introduces the concepts and mathematical relationships needed to study vehicle fuel consumption. It explains how fuel consumption is expressed, how the amount of fuel used can be estimated, and why driving speed is worth investigating.

### 2.1 Fuel consumption

For cars in many European countries, fuel consumption is commonly expressed in litres per 100 kilometres (L/100 km). This unit describes how many litres of fuel a vehicle uses, on average, to travel 100 kilometres.

A lower L/100 km value means that the vehicle uses less fuel to cover the same distance. For example, a vehicle with a consumption of 5 L/100 km uses less fuel over a 100-kilometre journey than a vehicle with a consumption of 8 L/100 km.

Fuel consumption is an average measure. The actual amount used can vary with driving conditions, so a single value should not be treated as a constant that applies to every situation.

### 2.2 Calculating the amount of fuel used

If the average fuel consumption and journey distance are known, the amount of fuel required can be estimated using:

$$F = \frac{D \times C}{100}$$

Where:

- $F$ is the estimated fuel used, in litres.
- $D$ is the distance travelled, in kilometres.
- $C$ is the average fuel consumption, in litres per 100 kilometres.

The division by 100 is necessary because the consumption value is given for every 100 kilometres.

For example, if a vehicle averages 6 L/100 km over a journey of 250 km, the estimated fuel used is:

$$F = \frac{250 \times 6}{100} = 15\text{ litres}$$

This calculation assumes that the average consumption value is representative of the whole journey.

### 2.3 Calculating fuel cost

Once the amount of fuel has been estimated, the fuel cost can be calculated by multiplying the fuel used by the price per litre:

$$K = F \times P$$

Where:

- $K$ is the estimated fuel cost, in euros.
- $F$ is the amount of fuel used, in litres.
- $P$ is the fuel price, in euros per litre.

For example, if a journey requires 15 litres of fuel and the assumed fuel price is €1.80 per litre, the estimated cost is €27.00.

The fuel price in this example is illustrative. Actual prices vary by location and over time.

### 2.4 Driving speed and fuel consumption

Driving speed is one factor that may influence a vehicle's fuel consumption. The relationship is not necessarily a simple straight line: fuel consumption can differ at low, moderate and high speeds, and the pattern depends on the vehicle and driving conditions.

Other factors can also affect fuel use, including acceleration, braking, road gradient, traffic, weather, vehicle load and engine condition.

For this project, we will focus on driving speed while keeping other conditions as consistent as possible in our data or assumptions. This will help us investigate the effect of speed without claiming that speed is the only factor that affects fuel consumption.

Research supports investigating this relationship. The U.S. Department of Energy summarises an Oak Ridge National Laboratory study of 74 vehicles, reporting that the best fuel economy was generally found around 40–50 mph and that fuel economy was about 27% lower at 80 mph than at 60 mph. These findings describe a broad trend across the vehicles studied; they do not provide measured consumption values for the hypothetical vehicle used in this project ([U.S. Department of Energy, 2022](https://www.energy.gov/cmei/vehicles/articles/fotw-1240-may-30-2022-slow-down-save-fuel-fuel-economy-decreases-about-27)).

The U.S. Environmental Protection Agency also explains that fuel economy varies with factors such as driving behaviour, vehicle condition, weather, road conditions and load ([U.S. EPA, *Your Mileage May Vary*](https://www.epa.gov/greenvehicles/your-mileage-may-vary)). This supports the decision to state simplifying assumptions and limitations clearly.

### 2.5 Mathematical modelling

## 5. Exploratory Data Analysis

This chapter examines the illustrative dataset before a mathematical model is fitted. Exploratory data analysis helps describe the values, identify patterns and check whether the data behave as expected. It uses summary statistics and a graph to investigate how the assumed fuel consumption changes with driving speed.

The analysis is descriptive: it shows the pattern in the five values created for this project. It does not establish the real-world fuel consumption of a particular car.

### 5.1 Descriptive Statistics

Descriptive statistics provide a compact summary of the fuel-consumption values. The mean gives the arithmetic average, the median is the middle value after sorting, and the range describes the difference between the largest and smallest values.

The following code calculates these statistics from the `fuel_data` DataFrame rather than entering the answers manually.

```python
mean_consumption = fuel_data["Fuel consumption (L/100 km)"].mean()
median_consumption = fuel_data["Fuel consumption (L/100 km)"].median()
minimum_consumption = fuel_data["Fuel consumption (L/100 km)"].min()
maximum_consumption = fuel_data["Fuel consumption (L/100 km)"].max()
consumption_range = maximum_consumption - minimum_consumption

print(f"Mean fuel consumption: {mean_consumption:.2f} L/100 km")
print(f"Median fuel consumption: {median_consumption:.2f} L/100 km")
print(f"Minimum fuel consumption: {minimum_consumption:.2f} L/100 km")
print(f"Maximum fuel consumption: {maximum_consumption:.2f} L/100 km")
print(f"Range: {consumption_range:.2f} L/100 km")
```

For this dataset, the mean fuel consumption is **5.82 L/100 km**, and the median is **5.50 L/100 km**. The lowest value is **5.00 L/100 km**, while the highest is **7.20 L/100 km**, giving a range of **2.20 L/100 km**.

The mean is slightly higher than the median because the larger consumption values at higher speeds raise the average. These statistics summarise the five assumed observations; they should not be interpreted as measured performance figures for a real vehicle.

### 5.2 Identifying the Lowest and Highest Consumption Values

It is useful to identify the speeds associated with the lowest and highest fuel-consumption values. This shows where the reference dataset reaches its minimum and maximum within the tested speed range.

```python
minimum_row = fuel_data.loc[fuel_data["Fuel consumption (L/100 km)"].idxmin()]
maximum_row = fuel_data.loc[fuel_data["Fuel consumption (L/100 km)"].idxmax()]

print(
    f"Lowest consumption: {minimum_row['Fuel consumption (L/100 km)']:.1f} L/100 km "
    f"at {minimum_row['Speed (km/h)']} km/h"
)
print(
    f"Highest consumption: {maximum_row['Fuel consumption (L/100 km)']:.1f} L/100 km "
    f"at {maximum_row['Speed (km/h)']} km/h"
)
```

The lowest assumed consumption occurs at **70 km/h**, with **5.0 L/100 km**. The highest occurs at **130 km/h**, with **7.2 L/100 km**.

This means that, within the five selected observations, 70 km/h is the most fuel-efficient of the tested speeds. It does not prove that 70 km/h is the most efficient speed for every vehicle or under all road conditions.

### 5.3 Visualising Speed and Fuel Consumption

A graph makes it easier to see how the two variables change together. The horizontal axis represents driving speed, while the vertical axis represents fuel consumption. Each marker represents one observation from the illustrative dataset.

The markers are connected to make the order of the observations easier to follow. The connecting lines are only a visual guide between the five data points; they do not mean that fuel consumption has been measured at every speed between them.

```python
plt.figure(figsize=(8, 5))
plt.plot(
    fuel_data["Speed (km/h)"],
    fuel_data["Fuel consumption (L/100 km)"],
    marker="o",
    linestyle="-"
)

plt.title("Illustrative Fuel Consumption at Different Driving Speeds")
plt.xlabel("Driving speed (km/h)")
plt.ylabel("Fuel consumption (L/100 km)")
plt.grid(True)
plt.xticks(fuel_data["Speed (km/h)"])
plt.tight_layout()
plt.show()
```

### 5.4 Interpreting the Graph

The graph shows a non-linear pattern in the illustrative data:

- From 50 km/h to 70 km/h, assumed fuel consumption decreases from 5.5 to 5.0 L/100 km.
- From 70 km/h onwards, assumed fuel consumption increases at each recorded speed.
- The increase becomes more noticeable at the higher speeds, reaching 7.2 L/100 km at 130 km/h.

The lowest point in the dataset is at 70 km/h. The values therefore suggest a U-shaped pattern over the selected speed range: consumption first decreases and then increases.

This pattern is useful when choosing a possible mathematical model in the next chapter. However, only five illustrative data points are available. A graph based on these points cannot show the exact behaviour between the recorded speeds, and the apparent pattern is partly determined by the values assumed when the dataset was created.

### 5.5 Comparing the Change Between Consecutive Speeds

The percentage change between two consecutive observations can help describe how strongly the assumed consumption changes as speed increases. It is calculated using:

\[
\text{Percentage change} = \frac{C_2-C_1}{C_1} \times 100
\]

Here, \(C_1\) is the first fuel-consumption value and \(C_2\) is the next value. A negative result represents a decrease, while a positive result represents an increase.

```python
speed_changes = fuel_data["Speed (km/h)"].diff()
consumption_changes = fuel_data["Fuel consumption (L/100 km)"].diff()
percentage_changes = fuel_data["Fuel consumption (L/100 km)"].pct_change() * 100

change_data = pd.DataFrame({
    "Speed interval (km/h)": [
        f"{int(fuel_data['Speed (km/h)'].iloc[i-1])}–{int(fuel_data['Speed (km/h)'].iloc[i])}"
        for i in range(1, len(fuel_data))
    ],
    "Change in consumption (L/100 km)": consumption_changes.iloc[1:].to_numpy(),
    "Percentage change (%)": percentage_changes.iloc[1:].to_numpy()
})

change_data.round(2)
```

The comparison shows that consumption decreases by approximately **9.09%** between 50 and 70 km/h. It then increases by approximately **6.00%**, **15.09%** and **18.03%** across the next three speed intervals.

These percentages describe changes between the selected data points only. They do not establish a continuous rate of change, and they should not be treated as experimental measurements.

### 5.6 Visualising Percentage Changes

The second graph displays the percentage change in fuel consumption between consecutive speed intervals. Negative bars indicate a decrease, while positive bars indicate an increase.

Unlike the first graph, which shows the fuel-consumption values themselves, this graph focuses on how much the values change from one observation to the next.

```python
plt.figure(figsize=(8, 5))

bar_colours = [
    "tab:red" if value < 0 else "tab:blue"
    for value in change_data["Percentage change (%)"]
]

plt.bar(
    change_data["Speed interval (km/h)"],
    change_data["Percentage change (%)"],
    color=bar_colours
)

plt.axhline(0, color="black", linewidth=0.8)
plt.title("Percentage Change in Fuel Consumption Between Speed Intervals")
plt.xlabel("Speed interval (km/h)")
plt.ylabel("Change in fuel consumption (%)")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

The graph shows a decrease of approximately **9.09%** between 50 and 70 km/h. The following intervals show increases of approximately **6.00%**, **15.09%** and **18.03%**.

The largest percentage increase occurs between 110 and 130 km/h. These values describe only the differences between the selected observations; they do not represent a continuous rate of change.

### 5.7 Limitations of the Exploratory Analysis

The dataset contains only five observations, and the fuel-consumption values are illustrative assumptions rather than measurements. There is no information about variation between repeated tests, measurement uncertainty or the behaviour at speeds that are not included.

Other influences on fuel consumption, such as traffic, wind, road gradient and vehicle load, are held constant by assumption rather than measured. The model also represents only one hypothetical vehicle.

These limitations mean that the data can demonstrate a mathematical modelling process, but they cannot establish the exact real-world fuel consumption of a specific car. Predictions made from a model fitted to these values should therefore be treated as illustrative estimates.
